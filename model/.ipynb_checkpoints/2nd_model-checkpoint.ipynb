```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("diabetes_type_800rows.csv")
```

```python
df
```

```python
df["Gender"] = df["Gender"].apply(lambda x: 1 if x == "Male" else 0)
```

```python
df = df.drop(columns=["PatientID","OutcomeBinary"])
```

```python
df["AutoimmuneMarker"] = df["AutoimmuneMarker"].apply(lambda x : 1 if x == "Yes" else 0)
```

```python
df
```

```python
df.DiabetesType.unique()
```

```python
X = df.drop(columns=["DiabetesType"])  # all columns except DiabetesType
y = df["DiabetesType"]  
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y_encoded,train_size=0.8,random_state=40)
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
standardized_data_train = scaler.transform(X_train)
```

```python
standardized_data_test = scaler.transform(X_test)
```

```python
classifier = svm.SVC(kernel='linear')
#training the support vector Machine Classifier
classifier.fit(X_train, y_train)
```

```python
# accuracy score on the training data
X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, y_train)
```

```python
training_data_accuracy
```

```python
X_test_predict = classifier.predict(X_test)
```

```python
training_data_accuracy = accuracy_score(X_test_predict, y_test)
```

```python
training_data_accuracy
```
